```python
import nltk

from nltk.corpus import wordnet

word = "book"
#word = "bank"

# Find meanings of the word
# Synset = Set of synonyms representing one particular meaning.
synsets = wordnet.synsets(word)
print(synsets)

print("Number of meanings:", len(synsets))

for synset in synsets:
    print("\nSynset:", synset)
    print("Definition:", synset.definition())
    print("Examples:", synset.examples())
```

```python
# synonyms

import nltk

from nltk.corpus import wordnet

word = "happy"

synonyms = set()

# Find synsets
for synset in wordnet.synsets(word):

    # Find words in each synset
    for lemma in synset.lemmas():

        synonyms.add(lemma.name())

print("Synonyms:")
print(synonyms)
```

```python
# Antonyms

import nltk

from nltk.corpus import wordnet

word = "good"

antonyms = set()

for synset in wordnet.synsets(word):

    for lemma in synset.lemmas():

        for antonym in lemma.antonyms():

            antonyms.add(antonym.name())

print("Antonyms:")
print(antonyms)
```

```python
# Hypernyms

import nltk

from nltk.corpus import wordnet

word = "dog"

synsets = wordnet.synsets(word)

for synset in synsets:

    print("\nSynset:", synset)

    hypernyms = synset.hypernyms()

    print("Hypernyms:")

    for hypernym in hypernyms:
        print(hypernym)
```

```python
# Hyponyms
import nltk

from nltk.corpus import wordnet

word = "animal"

synset = wordnet.synsets(word)[0]

print("Word:", synset)

print("\nHyponyms:")

for hyponym in synset.hyponyms():
    print(hyponym)
```

```python
#Meronym

from nltk.corpus import wordnet as wn

word = "car"

syn = wn.synsets(word)[0]

print("Word:", word)
print("Synset:", syn.name())
print("Definition:", syn.definition())

print("\nParts (Meronyms):")

for part in syn.part_meronyms():
    print(part.name())
```

```python
# Word Similarity

import nltk

from nltk.corpus import wordnet

# A higher value indicates greater similarity according to the Wu-Palmer similarity measure. (value between 0 and 1)

dog = wordnet.synsets("dog")[0]
cat = wordnet.synsets("cat")[0]

similarity = dog.wup_similarity(cat)

print("Similarity:", similarity)
```

```python
# Complete wordnet

import nltk

# Download only WordNet resources
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import wordnet

word = "happy"

print("Word:", word)

# Find meanings
synsets = wordnet.synsets(word)

print("\n--- Meanings ---")

for synset in synsets:

    print("\nSynset:", synset)
    print("Definition:", synset.definition())
    print("Examples:", synset.examples())


# Find synonyms
synonyms = set()

for synset in synsets:

    for lemma in synset.lemmas():

        synonyms.add(lemma.name())

print("\n--- Synonyms ---")
print(synonyms)


# Find antonyms
antonyms = set()

for synset in synsets:

    for lemma in synset.lemmas():

        for antonym in lemma.antonyms():

            antonyms.add(antonym.name())

print("\n--- Antonyms ---")
print(antonyms)
```

```python
# WordNet and Lemmatization

import nltk

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "cars",
    "studies",
    "playing",
    "better"
]

for word in words:

    print(word, "->", lemmatizer.lemmatize(word))
```
