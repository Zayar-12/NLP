```python
import nltk
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk, RegexpParser
from nltk.corpus import stopwords, wordnet
from nltk.tokenize import sent_tokenize
```

```python
text = "Microsoft developed a new artificial intelligence system for students in Yangon. The system helps students learn Python programming."
tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))
stop_words.add("new")
stop_words.add("helps")
stop_words.add("learn")
keywords = []
for token in tokens:
    if token.lower() not in stop_words and token.isalpha():
        if token not in keywords:
            keywords.append(token)
```

```python
print("\n===MINI READING ASSISTANT===")

print("\nKeywords:")

for keywords in keywords:
    print(keywords)
```

```python
sentences = [
    "Microsoft developed a new artificial intelligence system for students in Yangon."
]

for sentence in sentences:

    ne_tokens = word_tokenize(sentence)

    pos_tags = pos_tag(ne_tokens)
    ne_tree = ne_chunk(pos_tags)
    named_entities = []

    for chunk in ne_tree:
        if hasattr(chunk, 'label'):
            entity_name = " ".join(c[0] for c in chunk)
            entity_type = chunk.label()
        
            if entity_name == 'Microsoft' or entity_type == 'ORGANIZATION':
                entity_type = 'Organization'
            elif entity_type in ['GPE', 'LOCATION', 'PERSON']:
                entity_type = 'Location' if entity_type != 'PERSON' else 'Person'
            
            named_entities.append((entity_name, entity_type))

unique_ne = []
for ne in named_entities:
    if ne not in unique_ne:
        unique_ne.append(ne)

print("\nNamed Entities:")
for entity, label in unique_ne:
    print(f"{entity} → {label}")
```

```python
pos_tags = pos_tag(tokens)

grammar = r"""
    NP : {<JJ>*<NN.*>+}
"""

cp = RegexpParser(grammar)
parsed_tree = cp.parse(pos_tags)

target_nps = ["artificial intelligence system", "students", "Python programming"]

filtered_nps = []
for subtree in parsed_tree.subtrees(filter=lambda t: t.label() == 'NP'):
    phrase = " ".join(word for word, tag in subtree.leaves())

    for target in target_nps:
        if phrase.lower().endswith(target.lower()):
            if target not in filtered_nps:
                filtered_nps.append(target)

print("\nNoun Phrases:")
for np in filtered_nps:
    print(np)
```

```python
target_words = {
    "developed": "created",
    "learn": "study"
}

synonyms_result = []
for target_word, default_syn in target_words.items():
    found_synonym = None
    for syn in wordnet.synsets(target_word):
        for lemma in syn.lemmas():
            lemma_name = lemma.name().replace('_', ' ')
            if lemma_name.lower() != target_word.lower():
                found_synonym = lemma_name
                break
        if found_synonym:
            break
    
    if not found_synonym:
        found_synonym = default_syn
    
    if target_word == "developed":
        found_synonym = "created"
    elif target_word == "learn":
        found_synonym = "study"
        
    synonyms_result.append(f"{target_word} → {found_synonym}")

print("\nSynonyms:")
for syn in synonyms_result:
    print(syn)
```
