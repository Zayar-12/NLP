```python
# pip install summa

from summa import keywords

# Input text
text = """
Artificial intelligence is changing many industries.
Artificial intelligence is popular.
It is used in healthcare, education, finance, and transportation.
AI systems can analyze large amounts of data quickly.
Machine learning is an important part of artificial intelligence.
It allows computers to learn patterns from data.
AI can help people make better decisions and automate many tasks.
"""

# Extract keywords using TextRank
result = keywords.keywords(text, scores=True)

print("TextRank Keywords:")

for word, score in result:
    print(word, ":", round(score, 3))
```
