```python
# ==========================================
# Chunking / Shallow Parsing
# ==========================================

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
from nltk.tokenize import word_tokenize


# ------------------------------------------
# Required resources
# ------------------------------------------

# Run these if the resources are not installed
# nltk.download("punkt")
# nltk.download("punkt_tab")
# nltk.download("averaged_perceptron_tagger_eng")


# ------------------------------------------
# Input text
# ------------------------------------------

text = "The smart students study AI in the university."


# ------------------------------------------
# Step 1: Tokenization
# ------------------------------------------

# Split the text into individual tokens
tokens = word_tokenize(text)


# ------------------------------------------
# Step 2: POS Tagging
# ------------------------------------------

# Assign a POS tag to each token
tags = nltk.pos_tag(tokens)

print("POS Tags:")
print(tags)


# ------------------------------------------
# Step 3: Define Chunk Grammar
# ------------------------------------------

# NP = Noun Phrase
#
# <DT>?  → optional determiner
# <JJ>*  → zero or more adjectives
# <NN.*> → noun tags beginning with NN
#
# Examples:
# student
# the student
# smart student
# the smart student
#
# VP = Verb Phrase
#
# <VB.*> → any verb tag beginning with VB
# <NP>?  → optional noun phrase
#
# Examples:
# studies
# studies AI
# develops software
#
# PP = Prepositional Phrase
#
# <IN> → preposition
# <NP> → noun phrase
#
# Examples:
# in the university
# with the student

grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>}
    VP: {<VB.*><NP>?}
    PP: {<IN><NP>}
"""


# ------------------------------------------
# Step 4: Create Chunk Parser
# ------------------------------------------

# RegexpParser uses the grammar rules
# to identify and group chunks.
parser = nltk.RegexpParser(grammar)


# ------------------------------------------
# Step 5: Perform Chunking
# ------------------------------------------

# Apply the grammar to the POS-tagged text
tree = parser.parse(tags)


# ------------------------------------------
# Step 6: Display Chunk Tree
# ------------------------------------------

print("\nChunk Tree:")
print(tree)


# Display the tree visually
tree.pretty_print()
```
