```python

import nltk
from nltk import CFG
from nltk.parse import ChartParser
grammar = CFG.fromstring("""
    S -> NP VP
    NP -> Det N
    NP -> Det JJ N
    VP -> V NP PP ADVP
    PP -> P NP
    ADVP -> Adv
    Det -> 'the' | 'a'
    N -> 'teacher' | 'book' | 'library'
    Adv -> 'today'
    V -> 'reads'
    JJ -> 'clever'
    P -> 'in'
""")
```

```python
parser = nltk.ChartParser(grammar)

sentence = "the clever teacher reads a book in the library today"
words = sentence.split()

trees = list(parser.parse(words))

if trees:
    for tree in trees:
        print(tree)
        tree.pretty_print()  
else:
    print("No parse found.")
```
