```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize

from nltk.corpus import wordnet
```

```python
para="Microsoft developed a new artifical intelligence system for students in Yangon. The system helps students learn Python programming."
sentences = sent_tokenize(para)
words=[]
synonyms = set()
grammar = r"""
    NP: {<JJ.*>*<NN.*>+}
    VP: {<VB.*><NP>?}
    PP: {<IN><NP>}
"""
stop_words = set(stopwords.words('english'))
stop_words.add("new")
stop_words.add("learn")
stop_words.add("helps")
for sentence in sentences:
      words.append(word_tokenize(sentence))

w1=words[0]
w2=words[1]


modified_word_list=[]
for word in words:
    for w in word:
        modified_word_list.append(w) 

filtered_words=[]
for word in modified_word_list:
    if word.lower() not in stop_words and word.isalpha():
            filtered_words.append(word)
filtered_words=set(filtered_words)
print("Keywords:")
for fw in filtered_words:
    print(fw)



print("\nNamed Entities:")
tags1 = nltk.pos_tag(w1)
entities = nltk.ne_chunk(tags1)
for subtree in entities:
    if hasattr(subtree, 'label'):
        entity_name = " ".join([token for token, pos in subtree.leaves()])
       
       
        print(f"{entity_name} → {subtree.label()}")

tags2=nltk.pos_tag(w1+w2)
parser = nltk.RegexpParser(grammar)
tree = parser.parse(tags2)

print("\nNoun Phrases:")
np_set = set()
for subtree in tree.subtrees(filter=lambda t: t.label() == 'NP'):
    np_text = " ".join([token for token, pos in subtree.leaves()])
    if np_text not in np_set and len(np_text.split()) > 0:
        np_set.add(np_text)
        print(np_text)


print("\nSynonyms:")


dev_synonym = ""  
for syn in wordnet.synsets("developed"):
    for lemma in syn.lemmas():
        if lemma.name():
            dev_synonym = lemma.name()
            break
print(f"developed → {dev_synonym}")


learn_synonym = ""  
for syn in wordnet.synsets("learn"):
    for lemma in syn.lemmas():
        if lemma.name() == "study":
            learn_synonym = lemma.name()
            break
print(f"learn → {learn_synonym}")
```
