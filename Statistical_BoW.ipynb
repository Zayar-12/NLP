```python
# Import Library
# CountVectorizer automatically:	tokenizes text, creates vocabulary, counts words 

from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Create Sample Documents

documents = [
    "I love NLP",
    "I love AI",
    "NLP is interesting",
    "AI is powerful"
]

print(documents)
```

```python
# Create Bag of Words Model
# fit_transform() does two things:   1.	Learns vocabulary   2.	Creates vectors 

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)
X
```

```python
# Show Vocabulary
# Scikit-learn designed the default token_pattern requiring 2 or more characters so not include 'I'
print(vectorizer.get_feature_names_out())
```

```python
# Display BoW Matrix
print(X.toarray())
```

```python
# Convert Matrix into DataFrame

import pandas as pd

bow_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

bow_df
```

```python
# Test a New Sentence
new_sentence = [
    "It is a pretty cat"
]
```

```python
new_vector = vectorizer.transform(new_sentence)

print(new_vector.toarray())
```

```python
import pandas as pd

bow_df = pd.DataFrame(
    new_vector.toarray(),
    columns=vectorizer.get_feature_names_out()
)

bow_df
```
