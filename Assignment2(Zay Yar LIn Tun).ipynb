```python
import nltk
from nltk import CFG
from nltk.parse import ChartParser
```

```python

grammar = nltk.CFG.fromstring("""
    S -> NP VP

    NP -> Det N
    VP -> V NP

    Det -> 'the' | 'a'
    N -> 'student' | 'book' | 'teacher'
    V -> 'reads'
""")
```

```python
sentence1 = "the student reads a book."
sentence2 = "the student reads a"
sentence_arr= [sentence1, sentence2]
for sentence in sentence_arr:
 sentence = sentence.replace(".", "").lower()

 words = sentence.split()
 parser = ChartParser(grammar)
 trees = list(parser.parse(words))
 if trees:
    print("Sentence:",sentence)
    print("Result: Valid Sentence")
    print("Parse Trees: Generated")
    for tree in trees:
        print(tree)
        tree.pretty_print()  
else:
    print("Sentence:",sentence)
    print("Result: Invalid Sentence")
    print("No parse tree found.")
```
